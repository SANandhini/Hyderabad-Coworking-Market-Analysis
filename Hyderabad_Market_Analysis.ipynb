**Problem Statement**

The objective of this project is to analyze the coworking market in Hyderabad and identify suitable locations for business expansion.

**Data Collection**

Data was collected from publicly available sources including company websites, business directories, and Google Maps.

```python
import pandas as pd

df = pd.read_csv("Sample_coworking_data.csv")
```

**Data Cleaning**

```python
df.info()
df.isnull().sum()
```

**Exploratory Data Analysis**

Top locations

```python
df['Location'].value_counts()
```

Average price

```python
df.groupby('Location')['Price'].mean()
```

Seat capacity

```python
df['Seats'].describe()
```

**Visualizations**

```python
import matplotlib.pyplot as plt

df.groupby('Location')['Seats'].mean().plot(kind='bar')
plt.title("Average Seats by Location")
plt.show()
```

**Business Insights**

1. Hitech City has the highest concentration of coworking spaces.
2. Gachibowli shows strong demand and premium pricing.
3. Madhapur provides balanced pricing and accessibility.

**Recommendation**

Based on market density, pricing, and infrastructure availability, Hitech City and Gachibowli appear to be the strongest candidates for expansion.
